```python
import math

def check_winner(board):
    for i in range(3):
        if sum(board[i]) == 3:
            return 1
        if sum(board[i]) == -3:
            return -1
        
    for j in range(3):
        col_sum = board[j][0] + board[j][1] + board[j][2]

        if col_sum == 3:
            return 1
        if col_sum == -3:
            return -1
        
    diag1 = board[2][2] + board[1][1] + board[0][0]
    diag2 = board[2][0] + board[1][1] + board[0][2]

    if diag1 == 3 or diag2 == 3:
        return 1

    if diag1 == -3 or diag2 == -3:
        return -1
    

    return 0


def moves_left(board):
    for row in board:
        if 0 in row:
            return True
        
    return False

def minimax(board, depth, is_max):
    score = check_winner(board)

    if score == 1:
        return 1
    elif score == -1:
        return -1
    
    if not moves_left(board):
        return 0
    
    if is_max:
        best = -math.inf

        for i in range(3):
            for j in range(3):
                if(board[i][j] == 0):
                    board[i][j] = 1
                    value = minimax(board, depth+1, False)
                    board[i][j] = 0
                    best = max(best, value)

        return best
    else:
        best = math.inf

        for i in range(3):
            for j in range(3):
                if(board[i][j] == 0):
                    board[i][j] = -1
                    value = minimax(board, depth+1, True)
                    board[i][j] = 0
                    best = min(best, value)

        return best
    
def find_best_move(board, player):
    best_move = (-1, -1)

    if player == 1:
        best_val = -math.inf
        for i in range(3):
            for j in range(3):
                if(board[i][j] == 0):
                    board[i][j] = 1
                    value = minimax(board, 0, False)
                    board[i][j] = 0
                    if value > best_val:
                        best_move = (i, j)
                        best_val = value
    else:
        best_val = +math.inf
        for i in range(3):
            for j in range(3):
                if(board[i][j] == 0):
                    board[i][j] = -1
                    value = minimax(board, 0, True)
                    board[i][j] = 0
                    if value < best_val:
                        best_move = (i, j)
                        best_val = value


    return best_move
```

```python
def print_board(board):
    symbols = {1: 'X', -1: 'O', 0: ' '}
    print("\nBoard:")
    for row in board:
        print("|", end="")
        for cell in row:
            print(f" {symbols[cell]} |", end="")
        print()
    print()


def play_game():
    board = [[0, 0, 0],
             [0, 0, 0],
             [0, 0, 0]]

    print("Tic Tac Toe")
    print("You are Player B (O = -1)")
    print("AI is Player A (X = 1)")

    while True:
        print_board(board)

        # Check terminal state
        winner = check_winner(board)
        if winner == 1:
            print("AI (X) wins.")
            break
        elif winner == -1:
            print("You (O) win.")
            break
        elif not moves_left(board):
            print("Draw.")
            break

        # Human Move (Player B)
        row = int(input("Enter row (0-2): "))
        col = int(input("Enter col (0-2): "))

        if board[row][col] != 0:
            print("Invalid move. Try again.")
            continue

        board[row][col] = -1

        # Check after human move
        winner = check_winner(board)
        if winner != 0 or not moves_left(board):
            continue

        # AI Move
        ai_move = find_best_move(board, 1)
        board[ai_move[0]][ai_move[1]] = 1
        print(f"AI plays at position ({ai_move[0]}, {ai_move[1]})")


if __name__ == "__main__":
    play_game()
```
